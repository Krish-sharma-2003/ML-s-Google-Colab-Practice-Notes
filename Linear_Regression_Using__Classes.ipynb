```python
import numpy as np

class LinearRegression:
  def __init__(self, learning_rate=0.01, epochs=100):
    self.learning_rate = learning_rate
    self.epochs = epochs
    self.weights = 0
    self.bias = 0

  def predict(self, X):
    return self.weights * X + self.bias

  def compute_loss(self, y_true, y_pred):
    return np.mean((y_true - y_pred) ** 2)

  def fit(self, X, y):
      n = len(X)

      for i in range(self.epochs):
          y_pred = self.predict(X)
          dw = (-2/n) * np.sum(X * (y - y_pred))
          db = (-2/n) * np.sum(y - y_pred)


          self.weights -= self.learning_rate * dw
          self.bias -= self.learning_rate * db

          if i % 100 == 0:
            loss = self.compute_loss(y, y_pred)
            print(f"Epoch {i}, Loss: {loss:.4f},weight: {self.weights:.4f}, Bias:{self.bias:.4f}")


  def evaluate(self, X, y):
      y_pred = self.predict(X)
      loss = self.compute_loss(y, y_pred)
      print(f"FInal Loss: {loss:.4f}")

```

```python
X = np.array([1, 2, 3, 4, 5], dtype=np.float32)
y = np.array([2, 4, 6, 10, 12], dtype=np.float32)


X2 = np.array([1, 2, 3, 4, 5], dtype=np.float32)
y2 = np.array([3, 7, 8, 13, 14], dtype=np.float32)


#Create an object of our class
model = LinearRegression(learning_rate=0.01, epochs=2000)
model2 = LinearRegression(learning_rate=0.01, epochs=2000)


# Train the model
model.fit(X, y)
model2.fit(X2, y2)



#Evaluate the model
model.evaluate(X, y)
model.evaluate(X2, y2)


#Make a prediction
print("the Predicted model is: " , model.predict(8))
print("the Predicted model2 is: " , model2.predict(8))
```
